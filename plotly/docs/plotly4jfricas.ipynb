If the function below is not already in .fricas.input then define it here!
```
  sendHTML(s:String):Void ==
    html:=concat("--FORMAT:BEG:Formatted:0~%--FORMAT:BEG:HTML~%~A~%--",_
                   "FORMAT:END:HTML~%--FORMAT:END:Formatted:0~%")$String
    FORMAT('T,html,s)$Lisp
```
