```spad
(c1:=new(400,300)$CANVAS;_
c1.fillStyle "rgb(200 100 50)";_
c1.fillRect.(20, 40, 40, 60);_
c1.fillStyle  "rgb(0 100 200 / 50%)";_
c1.fillRect.(130, 130, 150, 150))
```

```spad
DF==> DoubleFloat
(c2:=new(400,300)$CANVAS;_
c2.beginPath;_
arc(c2,75::DF, 75.0::DF, 50.0::DF, 0::DF, 2*%pi::DF);_ 
moveTo(c2,110::DF, 75::DF);_
arcCC(c2,75::DF, 75, 35, 0, %pi);_
moveTo(c2,65::DF, 65);_
arc(c2,60, 65, 5, 0, 2*%pi);_
moveTo(c2,95, 65);_
arc(c2,90, 65, 5, 0, 2*%pi);_
c2.stroke)
```
